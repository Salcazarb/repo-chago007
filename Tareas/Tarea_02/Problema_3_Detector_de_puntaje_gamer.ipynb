```python
# números
num1 = int(input("Ingresa primer puntaje: "))
num2 = int(input("Ingresa segundo puntaje: "))

# suma
suma = num1 + num2

if suma < 50:
    print("\nResultado: Nivel principiante")
elif suma < 80:
    print("\nResultado: Nivel intermedio")
else:
    print("\nResultado: Nivel pro")
```
